## Testing connection

````python
from openai import OpenAI
client = OpenAI()

def get_response(prompt):
    response = client.chat.completions.create(
        model = "gpt-3.5-turbo",
        messages = [{"role": "user",
                     "content": prompt}],
        temperature = 0
    )
    return response.choices[0].message.content

prompt = "What is prompt engineering? Provide your answer in three simple bullet points."
response = get_response(prompt)
print(response)
````
